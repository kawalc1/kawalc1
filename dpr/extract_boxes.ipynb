```python
import cv2
from PIL import Image, ImageFilter
import matplotlib.pyplot as plt
import numpy as np
from scipy.cluster.hierarchy import linkage, fcluster
from scipy.spatial import distance as ssd

import imutils


def cv2_imshow(cv2image):
    plt.imshow(cv2.cvtColor(cv2image, cv2.COLOR_BGR2RGB))
    plt.show()

def unsharp_image(cv_image):
    image = Image.fromarray(cv_image)
    pil_im = image.filter(ImageFilter.UnsharpMask(radius=15, percent=350, threshold=3))
    # pil_im = image.filter(ImageFilter.UnsharpMask(radius=7, percent=150, threshold=2))
    sharpened_image = np.array(pil_im)
    return sharpened_image

def find_number_boxes(form_name: str):
    form_1_image = cv2.imread(form_name)
    imgray = cv2.cvtColor(form_1_image, cv2.COLOR_BGR2GRAY)
    
    unsharpened = unsharp_image(imgray)
    display(unsharpened)
    imgray = unsharpened
    
    kernel = np.ones((1,1),np.uint8)
    closed = cv2.morphologyEx(imgray, cv2.MORPH_CLOSE, kernel)
    cv2_imshow(closed)
    thresh = cv2.threshold(closed, 0, 255, cv2.THRESH_BINARY_INV | cv2.THRESH_OTSU)[1]
    display(Image.fromarray(thresh))
    # ret, thresh = cv2.threshold(imgray, 127, 255, 0)
    contours, hierarchy = cv2.findContours(thresh, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)
    
    
    cnts = []
    for c in contours:
        area = cv2.contourArea(c)
        x,y,w,h = cv2.boundingRect(c)
        ratio = w / h
        square_area = w * h
        
        squareness = area / square_area
        form_shape = form_1_image.shape
        # relative_size = form_shape[ /]
        
        
        if 0.7 < ratio < 1.0:
            print(area, square_area, squareness)
            # print(w, h)
            cnts.append(c)
    
    cv2.drawContours(form_1_image, cnts, -1, (0, 255, 0), 3)
    
    cv2_imshow(form_1_image)

find_number_boxes("forms/1.jpg")
find_number_boxes("forms/2.jpg")
find_number_boxes("forms/3.jpg")
find_number_boxes("forms/4.jpg")
find_number_boxes("forms/5.jpg")
find_number_boxes("forms/6.jpg")

```
